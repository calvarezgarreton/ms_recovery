```python
# clear workspace 
%reset -f

import pandas as pd
import numpy as np
import os

# === Load data ===
q_attrib = pd.read_csv('../../data/camels/catchment_attributes.csv')
q_attrib['p_mean_cr2met_1979_2010'] = 365*q_attrib['p_mean_cr2met_1979_2010'] 
gw_attrib = pd.read_csv('../../data/cr2sub/cr2sub_v1.1_attributes.csv')


id_basin_all  = q_attrib['gauge_id']
id_well_all = gw_attrib['cr2sub_id']

id_basin_ms  = pd.read_csv('../4_GW_Q_memory/csv/q_memory.csv')['gauge_id']
id_well_ms = pd.read_csv('../4_GW_Q_memory/csv/gw_memory.csv')['well_id']

n_basins_all = len(id_basin_all)
n_wells_all = len(id_well_all)

n_basins_ms = len(id_basin_ms)
n_wells_ms = len(id_well_ms)

basin_attrs = q_attrib[['gauge_id', 
                        'p_mean_cr2met_1979_2010', 
                        'aridity_cr2met_1979_2010',
                        'area_km2']].copy()

basin_attrs = basin_attrs[basin_attrs['gauge_id'].isin(id_basin_ms)]

print('Number of basins in CAMELS:', n_basins_all)
print('Number of wells in CR2SUB:', n_wells_all)
print('Number of basins in MS recovery:', n_basins_ms)
print('Number of wells in MS recovery:', n_wells_ms)    

print('Max precipitation (mm/yr) for basins in MS recovery:', round(basin_attrs['p_mean_cr2met_1979_2010'].max(), 0))   
print('Min precipitation (mm/yr) for basins in MS recovery:', round(basin_attrs['p_mean_cr2met_1979_2010'].min(), 0))   
print('Max aridity for basins in MS recovery:', round(basin_attrs['aridity_cr2met_1979_2010'].max(), 2))
print('Min aridity for basins in MS recovery:', round(basin_attrs['aridity_cr2met_1979_2010'].min(), 2))

print('Max area (km2) for basins in MS recovery:', round(basin_attrs['area_km2'].max(), 0))
print('Min area (km2) for basins in MS recovery:', round(basin_attrs['area_km2'].min(), 0))
print('Mean area (km2) for basins in MS recovery:', round(basin_attrs['area_km2'].mean(), 0))

```
